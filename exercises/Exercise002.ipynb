## Reading a FASTA file

Write a parser thats reads and processes a FASTA file called `ecoli_sequences.fasta` into a list named `ecoli`. Repeat the process for the files `archaea_sequences.fasta` and `plant_sequences.fasta`. Make sure to name the resulting variables appropriately to ensure clean code principles.

-----
**Tips**

> * FASTA files follow a stringent structure: Header first, followed by the actual sequence. Inspect the file using print to find a pattern.
> * Try using functions to solve the problem. While not mandatory, it can save code lines and generalize a concept such as a FASTA parser. Click [here](https://www.w3schools.com/python/python_functions.asp) for more on Python functions.

**FASTA Schema**

```
0 | >Header1
1 | Sequence1
2 | >Header2
3 | Sequence2
```

````python
def parse_fasta(file_path):
    """Parses a FASTA file into a list of dictionaries with headers and sequences."""
    sequences = []
    current_header = None

    with open(file_path, "r") as file:
        for line in file:
            line = line.strip()
            if not line:
                continue  # Skip empty lines if any

            if line.startswith(">"):
                # Clean up the '>' symbol from the header name
                current_header = line[1:]
            else:
                # If we have a header, pair it with this sequence line
                if current_header is not None:
                    sequences.append({
                        "header": current_header,
                        "sequence": line
                    })
    return sequences

# Parse all three downloaded files
ecoli = parse_fasta("ecoli_sequences.fasta")
archaea = parse_fasta("archaea_sequences.fasta")
plants = parse_fasta("plant_sequences.fasta")

# Print the count of sequences found to verify it worked
print(f"Loaded {len(ecoli)} E. coli, {len(archaea)} Archaea, and {len(plants)} plant sequences.")
````

````python
def write_fasta(file_path, datasets_dict):
    """
    Writes multiple sequence datasets to a single FASTA file.
    datasets_dict should map an organism name to its sequence list.
    """
    with open(file_path, "w") as file:
        for organism, sequences in datasets_dict.items():
            for idx, entry in enumerate(sequences, start=1):
                # Construct a unique, descriptive header
                new_header = f">{organism}_ID{idx}_{entry['header']}\n"
                sequence_line = f"{entry['sequence']}\n"

                file.write(new_header)
                file.write(sequence_line)

# Combine datasets into a dictionary mapping organism names to their respective lists
data_to_write = {
    "Ecoli": ecoli,
    "Archaea": archaea,
    "Plant": plants
}

# Execute the writing function
write_fasta("all_sequences.fasta", data_to_write)
print("Successfully merged and saved all sequences to 'all_sequences.fasta'!")
````
